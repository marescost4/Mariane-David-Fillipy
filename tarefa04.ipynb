# Conceitos Teóricos


---


## Mediana

É a terceira medida de tendência
central, sendo definida como o valor que divide uma série ordenada de tal forma que pelo menos metade ou 50% dos itens sejam iguais ou maiores que ela, dividindo o conjunto em partes iguais.


---


## Quartis

São em sua essência medidas de posição semelhantes a mediana, entretanto não são medidas de tendência central. Enquanto a média divide um grupo em duas partes iguais, os Quartis fazem essa divisão em 4, possuindo 3 separatrizes, chamados de Quartis (Q1, Q2, Q3).
Dados um conjunto ordenado de valores em ordem crescente, o primeiro quartil (Q1) é o valor que divide esse conjunto em ¼ ou 25% e ¾ ou 75%, sendo uma divisão de partes distintas.
Já o segundo quartil (Q2) é o valor que divide esse grupo em duas partes semelhantes (50/50), por isso esse quartil também é chamado de mediana.
Dados esse mesmo conjunto, o terceiro quartil (Q3) divide o grupo em duas partes, no qual os números menores que esse quartil representam ¾ ou 75% do total, enquanto os números maiores representam ¼ ou 25% do total, sendo o oposto do Q2.

## Amplitude Interquartílica

É uma medida de variação dada pela diferença entre os valores do terceiro e primeiro quartil de um conjunto de dados; representam o comprimento do intervalo que contém os 50% internos do conjunto.

## Observações Atípicas

Observações nas extremidades de uma amostra, que estão tão distantes do corpo principal dos dados que a possibilidade de incluí-la na amostra é questionável, pois podem gerar grandes erros em análises de dados.

## Boxplot

É um gráfico estatístico usado para descrever a distribuição de dados numéricos, resumindo informações centrais como a mediana, os quartis e os valores mínimo e máximo, auxiliando também na observação de pontos de observação atípica.

# Como determinamos se há observações atípicas em uma base de dados?

Podemos determinar a presença dessas observações atípicas através do método dda Amplitude interquartílica.

Primeiramente é necessário organizar os dados em ordem crescente, para que seja possivel determinar os quartis, no qual os 25% menores dados serão separados pelo Q1, a Q2 vai separar os dados totais em 50% e o Q3 também separa o valor total, mas em 75%.

Em seguida a amplitude Interquartílica é calculada através da seguinte equação:

IQR = Q3 - Q1

no qual IQR representa a amplitude Interquartílica; Q3 representa o terceiro quartil e Q1 o primeiro. Esse cálculo representa a dispersão dos 50% centrais dos dados.

Em seguida, é necessário determinar os limites para observações atípicas utilizando as equaçõs a seguir:

Limite inferior = Q1 - 1,5(IQR)
Limite superior = Q3 + 1,5(IQR)

então quaqualquer numero menor que o limite inferior e maior que o limite superior, é uma observação atípica.
