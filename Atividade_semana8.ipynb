A. Um pesquisador verifica se uma determinada semente germinou ou não.

Bernoulli, porque esse conceito apresenta apenas dois desfechos possiveis, por exemplo: Fracasso(não germinar) ou sucesso(germinar)

B. De 20 sementes plantadas, deseja-se calcular a probabilidade de exatamente 15
germinarem, considerando que a probabilidade de germinação de cada semente é 0,8.

Utilizamos o método Binomial, no que consiste em determinar o número de sucessos em um número fixo de ensaios de Bernoulli.

Para isso utilizamos a fórmula a seguir:

X~Bin(n,p)

$$P(X = k) = \binom{n}{k} p^k (1 - p)^{n - k}$$

onde:

$n = 20$ (número total de sementes)

$k = 15$ (número de sementes germinadas desejado)

$p = 0{,}8$ (probabilidade de germinação)

$1 - p = 0{,}2$ (probabilidade de não germinação)

Assim:

$$\binom{20}{15} = \frac{20!}{15! \cdot (20 - 15)!} = \frac{20!}{15! \cdot 5!}$$


$$\binom{20}{15} = \frac{20 \times 19 \times 18 \times 17 \times 16}{5 \times 4 \times 3 \times 2 \times 1} = \frac{1.860.480}{120} = 15.504$$

$$P(X = 15) = 15.504 \times (0{,}8)^{15} \times (0{,}2)^5$$

$$P(X = 15) = 15.504 \times 0{,}03518437 \times 0{,}00032$$

$$P(X = 15) \approx 0{,}17456$$

então a probabilidade de especificamente 15 sementes germinarem, é de aproximadamente 0,175.

C. Em uma área de mata, são observados, em média, 4 ninhos de uma determinada
espécie por hectare. Deseja-se calcular a probabilidade de encontrar exatamente 6 ninhos em um hectare.

É necessário utilizar o modelo de Poisson, que é utilizado para representar o número de vezes que determinado evento ocorre em um intervalo definido de tempo, espaço, área, volume ou outra unidade de observação (nesse caso, hectare).
Para isso, utilizaremos a seguinte fórmula:

$$P(X = k) = \frac{e^{-\lambda} \cdot \lambda^k}{k!}$$

Onde:

$\lambda = 4$ (taxa média de ninhos por hectare)

$k = 6$ (quantidade de ninhos que desejamos encontrar)

$e \approx 2{,}71828$ (base do logaritmo natural)

$$P(X = 6) = \frac{e^{-4} \cdot 4^6}{6!}$$

Assim temos que:

Potência de e:
    
e−4≈0,0183156

Potência de λ:
    
46=4.096

Fatorial de k:
    
6!=6×5×4×3×2×1=720

Por fim calculamos:

$$P(X = 6) = \frac{0{,}0183156 \times 4.096}{720}$$

$$P(X = 6) = \frac{75{,}0207}{720} \approx 0{,}10419$$

A probabilidade de encontrar exatamente 6 ninhos em um hectare é de aproximadamente 10,42% (ou 0,1042).

Questões

a) Qual distribuição deve ser utilizada? Explique por quê.

Deve ser utilizado a distribuição binomial, pois o número total de ensaios já estão definidos e é necessario repetir vários testes de Bernoulli independentes com a mesma chance de acerto.

b) Qual é a probabilidade de exatamente 3 abelhas apresentarem a característica?

Usamos a fórmula da Distribuição Binomial onde inserimos os dados:

$n = 10$

$k = 3$

$p = 0{,}30$

$1 - p = 0{,}70$

$$\binom{10}{3} = \frac{10!}{3! \cdot (10 - 3)!} = \frac{10 \times 9 \times 8}{3 \times 2 \times 1} = 120$$

Calculamos as potências

$(0{,}30)^3 = 0{,}027$

$(0{,}70)^7 \approx 0{,}0823543$

Assim:

$$P(X = 3) = 120 \times 0{,}027 \times 0{,}0823543$$

$$P(X = 3) \approx 0{,}266828$$

Então, descobrimos que a probabilidade de exatamente três abelhas apresentarem a característica é de aproximadamente 0,267.


c) Qual é a probabilidade de pelo menos 1 das 10 abelhas apresentar a característica?

Para calcular utilizaremos o evento complementar

Onde: $$P(X \ge 1) = 1 - P(X = 0)$$ que define a probabilidade de pelo menos 1 desses individuos apresentar a característica.

Assim definimos:

$$P(X = 0) = \binom{10}{0} (0{,}30)^0 (0{,}70)^{10}$$

$$P(X = 0) = 1 \times 1 \times (0{,}70)^{10} \approx 0{,}0282475$$

Dessa forma, podemos observar que a probabilidade de pelo menos 1 dos indivíduos apresentar essa característica é de aproximadamente 0,028.

3. Na observação ecológica (10 min)
Durante levantamentos em uma área de cerrado, pesquisadores registram em média 2
encontros com uma determinada espécie de mamífero por hora.
Questões

a) Qual distribuição seria mais adequada para assumir nesse caso?

A mais adequada seria a distribuição de Poisson.

b) Qual é a probabilidade de ocorrerem exatamente 3 encontros em uma hora?

$$P(X = k) = \frac{e^{-\lambda} \cdot \lambda^k}{k!}$$

Onde:

$\lambda = 2$

$k = 3$

$e \approx 2{,}71828$ (base do logaritmo natural)

Então:

$$P(X = k) = \frac{2{,}71828^{-2} \cdot 2^3}{2!}$$

Assim, calculamos os expoentes:

$$P(X = k) = \frac{0{,}135 \cdot 8}{2}$$

Assim temos o resultado de que

$$P(X = k) = \frac{0{,}135 \cdot 8}{2} = 0,54$$

Ou seja, a probabilidade é de aproximadamente 0,54.

c) Qual é a probabilidade de nenhum encontro ocorrer durante uma hora?

$$P(X = k) = \frac{e^{-\lambda} \cdot \lambda^k}{k!}$$

Onde:

$\lambda = 2$

$k = 0$

$e \approx 2{,}71828$ (base do logaritmo natural)

Então:

$$P(X = k) = \frac{2{,}71828^{-2} \cdot 2^0}{2!}$$

Assim, calculamos os expoentes:

$$P(X = k) = \frac{0{,}135 \cdot 1}{2}$$

Assim temos o resultado de que

$$P(X = k) = \frac{0{,}135 \cdot 1}{2} = 0,0675$$

Ou seja, a probabilidade é de aproximadamente 0,0675.

d) Qual é a probabilidade de ocorrerem mais de 2 encontros em uma hora?

$$P(X > 2) = 1 - P(X \le 2)$$

$$P(X > 2) = 1 - \left[ P(X = 0) + P(X = 1) + P(X = 2) \right]$$

Assim calculamos a probabilidade de ocorrer 0, 1 e 2 encontros respectivamente

Para $k = 0$ (nenhum encontro):$$P(X = 0) = \frac{e^{-2} \cdot 2^0}{0!} = \frac{0{,}135335 \times 1}{1} = 0{,}135335$$Para $k = 1$ (exatamente 1 encontro):$$P(X = 1) = \frac{e^{-2} \cdot 2^1}{1!} = \frac{0{,}135335 \times 2}{1} = 0{,}270670$$Para $k = 2$ (exatamente 2 encontros):$$P(X = 2) = \frac{e^{-2} \cdot 2^2}{2!} = \frac{0{,}135335 \times 4}{2} = 0{,}270670$$

Depois, somamos os resultados e subtraímos pelo total(1):

$$P(X \le 2) = P(X = 0) + P(X = 1) + P(X = 2)$$$$P(X \le 2) = 0{,}135335 + 0{,}270670 + 0{,}270670 = 0{,}676675$$

$$P(X > 2) = 1 - 0{,}676675 = 0{,}323325$$

Assim, a probabilidade de ocorrerem mais de 2 encontros em uma hora é de aproximadamente 32,33% (ou 0,3233).
