```python
import pandas as pd
df = pd.read_excel('Dados_Tarefa03.xlsx')
df.head (21)
```

```python
frequencia_sexo = df['Sexo'].value_counts()
display(frequencia_sexo)
```

Esta tabela descreve a frequência de estudantes por sexo, sendo M para masculino e F para feminino. Como resultado o número de estudantes do sexo masculino demonstra uma maior proporção.  

```python
frequencia_horas_estudo = df['Horas de estudo em biologia/semana'].value_counts().sort_index()
frequencia_horas_estudo_df = frequencia_horas_estudo.reset_index()
frequencia_horas_estudo_df.columns = ['Horas de estudo em biologia/semana', 'numero de estudantes']
display(frequencia_horas_estudo_df)
```

Esta tabela demonstra que as horas de estudo semanais dos estudantes variaram de 1 a 39. As maiores frequências foram observadas entre os estudantes que estudavam 4, 26 e 33 horas por semana.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

sns.set_theme(style="ticks")

f, ax = plt.subplots(figsize=(10, 6))
sns.despine(f)
sns.histplot(df,x="Horas de estudo em biologia/semana", kde=True, color="skyblue",edgecolor=".3",linewidth=.5)
ax.set_title('Distribuição das Horas de Estudo em Biologia por Semana')
ax.set_xlabel('Horas de Estudo por Semana')
ax.set_ylabel('Frequência')
plt.xticks(np.sort(df['Horas de estudo em biologia/semana'].unique()))
plt.tight_layout()
plt.show()
```

Este histograma mostra a distribuição das horas de estudo por semana em biologia para cada estudante. Isso permite observar a frequência de períodos semelhantes entre os estudantes.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.scatterplot(data=df,x='Estudante', y='Horas de estudo em biologia/semana', hue='Sexo',s=100, alpha=0.7)
plt.title('Relação entre Horas de Estudo em Biologia e Estudante por Sexo')
plt.xlabel('Quantidade de estudantes')
plt.ylabel('Quantidade de horas')
plt.xticks(df['Estudante'].unique())
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

Este gráfico de dispersão relaciona as horas de estudo em biologia para cada estudante, contendo pontos coloridos de acordo com o sexo (F para Feminino, M para Masculino). Isso permite observar a distribuição das horas de estudo entre os estudantes e identificar se há alguma diferença aparente entre os grupos de sexo.

```python

media_horas_sexo = df.groupby('Sexo')['Horas de estudo em biologia/semana'].mean().reset_index()
plt.figure(figsize=(8, 6))
sns.barplot(data=media_horas_sexo, x='Sexo', y='Horas de estudo em biologia/semana', hue='Sexo', palette='muted', legend=False)
plt.title('Média de Horas de Estudo por Sexo')
plt.xlabel('Sexo')
plt.ylabel('Média de Horas de Estudo')
plt.grid(axis='y', linestyle='--', alpha=0.7)
for index, row in media_horas_sexo.iterrows():plt.text(index, row['Horas de estudo em biologia/semana'] + 0.5, round(row['Horas de estudo em biologia/semana'], 2),color='black', ha='center')
plt.tight_layout()
plt.show()
```

Este gráfico de barras mostra a média de horas de estudo em biologia em relação as barras coloridos de acordo com o sexo (F para Feminino, M para Masculino). Isso permite observar a distribuição da média de horas de estudo entre os sexos.
