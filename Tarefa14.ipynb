##Exercício 1 — Massa corporal de mamíferos

A massa corporal de uma espécie de pequeno mamífero apresenta distribuição normal:

• μ=420 g

• σ=55 g

Determine:

Para resolver as questões, padronizamos a variável $X \sim \mathcal{N}(420,\, 55^2)$ para a distribuição normal padrão $Z \sim \mathcal{N}(0, 1)$ utilizando a fórmula:$$Z = \frac{X - \mu}{\sigma} = \frac{X - 420}{55}$$



1. A probabilidade de um indivíduo pesar menos de 350 g.

$$Z = \frac{350 - 420}{55} = \frac{-70}{55} \approx -1{,}27$$


$$P(Z < -1{,}27) \approx 0{,}1020$$


Resposta: 10,20% de probabilidade (ou $0{,}1020$).

2. A probabilidade de pesar mais de 500 g.


$$Z = \frac{500 - 420}{55} = \frac{80}{55} \approx 1{,}45$$


Usando a propriedade do complementar:

$$P(Z > 1{,}45) = 1 - P(Z < 1{,}45) = 1 - 0{,}9265 = 0{,}0735$$Resposta: 7,35% de probabilidade (ou $0{,}0735$).


3. A probabilidade de pesar entre 400 e 480 g.


Calculando os valores de $Z$:


Para $X_1 = 400$: $Z_1 = \frac{400 - 420}{55} = \frac{-20}{55} \approx -0{,}36$


Para $X_2 = 480$: $Z_2 = \frac{480 - 420}{55} = \frac{60}{55} \approx 1{,}09$

Buscando os valores acumulados na tabela:

$P(Z < 1{,}09) \approx 0{,}8621$

$P(Z < -0{,}36) \approx 0{,}3594$

$P(-0{,}36 < Z < 1{,}09) = 0{,}8621 - 0{,}3594 = 0{,}5027$

Resposta: 50,27% de probabilidade (ou $0{,}5027$).


```python
#4. Represente graficamente as três probabilidades
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Parâmetros da distribuição normal
mu = 420      # Média
sigma = 55    # Desvio padrão

# 2. Vetor de valores de X (de mu - 4*sigma até mu + 4*sigma)
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = norm.pdf(x, mu, sigma)

# 3. Criar a figura com 3 subplots
fig, axes = plt.subplots(3, 1, figsize=(10, 12), sharex=True)

# Define as cores
color_curve = '#1f77b4'
color_p1 = '#e74c3c'  # Vermelho
color_p2 = '#2ecc71'  # Verde
color_p3 = '#f39c12'  # Laranja

# -----------------------------------------------------------------------------
# Gráfico 1: P(X < 350)
# -----------------------------------------------------------------------------
ax1 = axes[0]
ax1.plot(x, y, color=color_curve, lw=2)

x_fill1 = np.linspace(mu - 4 * sigma, 350, 500)
ax1.fill_between(x_fill1, norm.pdf(x_fill1, mu, sigma), color=color_p1, alpha=0.5,
                 label=r'$P(X < 350) \approx 10,20\%$')

ax1.axvline(350, color=color_p1, linestyle='--', lw=1.5)
ax1.axvline(mu, color='black', linestyle=':', alpha=0.6, label=r'$\mu = 420\text{ g}$')
ax1.set_title(r'1. $P(X < 350\text{ g})$ — Pesar menos de 350 g', fontsize=12, fontweight='bold')
ax1.set_ylabel('Densidade de Probabilidade')
ax1.legend(loc='upper right')
ax1.grid(True, alpha=0.3)

# -----------------------------------------------------------------------------
# Gráfico 2: P(X > 500)
# -----------------------------------------------------------------------------
ax2 = axes[1]
ax2.plot(x, y, color=color_curve, lw=2)

x_fill2 = np.linspace(500, mu + 4 * sigma, 500)
ax2.fill_between(x_fill2, norm.pdf(x_fill2, mu, sigma), color=color_p2, alpha=0.5,
                 label=r'$P(X > 500) \approx 7,35\%$')

ax2.axvline(500, color=color_p2, linestyle='--', lw=1.5)
ax2.axvline(mu, color='black', linestyle=':', alpha=0.6)
ax2.set_title(r'2. $P(X > 500\text{ g})$ — Pesar mais de 500 g', fontsize=12, fontweight='bold')
ax2.set_ylabel('Densidade de Probabilidade')
ax2.legend(loc='upper right')
ax2.grid(True, alpha=0.3)

# -----------------------------------------------------------------------------
# Gráfico 3: P(400 < X < 480)
# -----------------------------------------------------------------------------
ax3 = axes[2]
ax3.plot(x, y, color=color_curve, lw=2)

x_fill3 = np.linspace(400, 480, 500)
ax3.fill_between(x_fill3, norm.pdf(x_fill3, mu, sigma), color=color_p3, alpha=0.5,
                 label=r'$P(400 < X < 480) \approx 50,27\%$')

ax3.axvline(400, color=color_p3, linestyle='--', lw=1.5)
ax3.axvline(480, color=color_p3, linestyle='--', lw=1.5)
ax3.axvline(mu, color='black', linestyle=':', alpha=0.6)
ax3.set_title(r'3. $P(400 < X < 480\text{ g})$ — Pesar entre 400 g e 480 g', fontsize=12, fontweight='bold')
ax3.set_xlabel('Massa Corporal (g)')
ax3.set_ylabel('Densidade de Probabilidade')
ax3.legend(loc='upper right')
ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

##Exercício 2 — Percentis de tamanho corporal

O comprimento corporal desses mamíferos segue uma distribuição normal com:

• μ=150 cm

• σ=10,5 cm

Calcule:
1. O percentil 80.


O percentil 80 é o valor $x$ tal que a área acumulada à esquerda é $0{,}80$
($P(X \le x) = 0{,}80$).

Consultando a tabela $Z$ inversa (ou score $Z$ acumulado de $0{,}80$): $Z \approx 0{,}84$ (mais precisamente $0{,}8416$). Convertendo para a escala original ($X$):

$$X = 150 + (0{,}8416 \cdot 10{,}5) = 150 + 8{,}84 = 158{,}84\text{ cm}$$

Interpretação Biológica:

Um comprimento de $158{,}84\text{ cm}$ separa os $80\%$ menores mamíferos da população dos $20\%$ maiores. Ou seja, $80\%$ dos indivíduos desta espécie medem até $158{,}84\text{ cm}$.


2. O valor correspondente a 10% dos indivíduos apresentarem o comprimento superior a esse
valor.


Se $10\%$ dos indivíduos apresentam comprimento superior a esse valor, então $90\%$ estão abaixo dele ($P(X \le x) = 0{,}90$), o que corresponde ao Percentil 90 ($P_{90}$).

Consultando a tabela $Z$ inversa para área acumulada de $0{,}90$:

$Z \approx 1{,}28$ (mais precisamente $1{,}2816$).

Convertendo para a escala original ($X$):


$$X = 150 + (1{,}2816 \cdot 10{,}5) = 150 + 13{,}46 = 163{,}46\text{ cm}$$


Interpretação Biológica: O comprimento de $163{,}46\text{ cm}$ limita a cauda superior da população. Apenas os $10\%$ maiores mamíferos da espécie possuem comprimento acima de $163{,}46\text{ cm}$. Em termos ecológicos ou de seleção natural, este limiar pode definir indivíduos adultos dominantes ou de maior aptidão física (fitness).

```python
# 3. Faça um único gráfico indicando os três percentis.
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parâmetros
mu = 150
sigma = 10.5

# Cálculos dos percentis
p50 = norm.ppf(0.50, loc=mu, scale=sigma) # 150.00 cm
p80 = norm.ppf(0.80, loc=mu, scale=sigma) # 158.84 cm
p90 = norm.ppf(0.90, loc=mu, scale=sigma) # 163.46 cm

# Eixo X e curva de densidade
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = norm.pdf(x, mu, sigma)

plt.figure(figsize=(10, 6))
plt.plot(x, y, color='#1f77b4', lw=2.5, label=r'Distribuição Normal: $\mathcal{N}(150; 10{,}5^2)$')

# Destaques das linhas dos percentis
plt.axvline(p50, color='#7f7f7f', linestyle=':', lw=2, label=f'P50 (Mediana) = {p50:.1f} cm')
plt.axvline(p80, color='#f39c12', linestyle='--', lw=2, label=f'P80 = {p80:.2f} cm (80% menores)')
plt.axvline(p90, color='#e74c3c', linestyle='--', lw=2, label=f'P90 = {p90:.2f} cm (10% maiores)')

# Sombreados das regiões
plt.fill_between(x, y, where=(x <= p80), color='#f39c12', alpha=0.15)
plt.fill_between(x, y, where=(x >= p90), color='#e74c3c', alpha=0.3, label=r'10% Maiores ($X > 163{,}46\text{ cm}$)')

plt.title('Distribuição Normal do Comprimento Corporal dos Mamíferos', fontsize=13, fontweight='bold', pad=15)
plt.xlabel('Comprimento Corporal (cm)', fontsize=11)
plt.ylabel('Densidade de Probabilidade', fontsize=11)
plt.legend(loc='upper left', frameon=True)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

Para compor o gráfico com três percentis, incluímos também o Percentil 50 ($P_{50}$ / Mediana), que coincide com a média ($\mu = 150\text{ cm}$):

$P_{50}$ (Mediana): $150{,}00\text{ cm}$ (divide a população exatamente ao meio, $50\%$ menores e $50\%$ maiores).

$P_{80}$: $158{,}84\text{ cm}$ ($80\%$ menores).$P_{90}$: $163{,}46\text{ cm}$ (limite para os $10\%$ maiores).

##Exercício 3 — Integração
Uma população de aves apresenta massa corporal normalmente distribuída com:


• μ=72 g


• σ=8 g


Em Python:


1. Calcule P(X≤60).
2. Calcule P(X≥85).
3. Calcule P(65≤X≤80).
4. Calcule os percentis 90, 95 e 99.
5. Produza um gráfico da distribuição normal mostrando as três regiões de probabilidade e os
três percentis.
6. Escreva uma interpretação ecológica curta dos resultados.

```python
#1. Calcule P(X≤60).

from scipy.stats import norm

# Parâmetros da distribuição
mu = 72     # Média (g)
sigma = 8   # Desvio padrão (g)

# Cálculo de P(X <= 60) usando a função de distribuição acumulada (CDF)
probabilidade = norm.cdf(60, loc=mu, scale=sigma)

# Exibição do resultado
print(f"Probabilidade: {probabilidade:.6f}")
print(f"Em percentagem: {probabilidade * 100:.2f}%")
```

```python
#2. Calcule P(X≥85).

from scipy.stats import norm

# Parâmetros da distribuição
mu = 72     # Média (g)
sigma = 8   # Desvio padrão (g)

# Opção 1: Usando a Survival Function / Cauda Superior (sf)
probabilidade = norm.sf(85, loc=mu, scale=sigma)

# Opção 2: Usando o complementar da CDF (1 - CDF)
# probabilidade = 1 - norm.cdf(85, loc=mu, scale=sigma)

# Exibição do resultado
print(f"Probabilidade: {probabilidade:.6f}")
print(f"Em percentagem: {probabilidade * 100:.2f}%")
```

```python
#3. Calcule P(65≤X≤80).
from scipy.stats import norm

# Parâmetros da distribuição
mu = 72     # Média (g)
sigma = 8   # Desvio padrão (g)

# Cálculo de P(65 <= X <= 80) = CDF(80) - CDF(65)
probabilidade = norm.cdf(80, loc=mu, scale=sigma) - norm.cdf(65, loc=mu, scale=sigma)

# Exibição do resultado
print(f"Probabilidade: {probabilidade:.6f}")
print(f"Em percentagem: {probabilidade * 100:.2f}%")
```

```python
#4. Calcule os percentis 90, 95 e 99.
from scipy.stats import norm

# Parâmetros da distribuição
mu = 72     # Média (g)
sigma = 8   # Desvio padrão (g)

# Cálculo dos percentis 90, 95 e 99
p90 = norm.ppf(0.90, loc=mu, scale=sigma)
p95 = norm.ppf(0.95, loc=mu, scale=sigma)
p99 = norm.ppf(0.99, loc=mu, scale=sigma)

# Exibição dos resultados
print(f"Percentil 90 (P90): {p90:.2f} g")
print(f"Percentil 95 (P95): {p95:.2f} g")
print(f"Percentil 99 (P99): {p99:.2f} g")
```

```python
#5. Produza um gráfico da distribuição normal mostrando as três regiões de probabilidade e os três percentis.
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# --- Definição dos Parâmetros do Exercício ---
mu = 72     # Média (g)
sigma = 8   # Desvio padrão (g)

# --- Definição das Regiões de Probabilidade (Limites) ---
limit1 = 60 # P(X <= 60) -> Cauda Inferior
limit2 = 85 # P(X >= 85) -> Cauda Superior

# --- Definição dos Percentis (P90, P95, P99) ---
percentiles = [0.90, 0.95, 0.99]
# Calcula os valores de X correspondentes aos percentis usando ppf
x_percentiles = norm.ppf(percentiles, loc=mu, scale=sigma)

# --- Definição do Eixo X e Curva de Densidade N(72, 8) ---
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = norm.pdf(x, mu, sigma)

# --- Criação do Gráfico (ÚNICA FIGURE INTEGRADA) ---
plt.figure(figsize=(12, 7))
plt.plot(x, y, color='#1f77b4', lw=3, label=r'Distribuição Normal N(72, 8)')

# --- Sombreado das Regiões de Probabilidade ---
# Região 1: X <= 60 g (Cauda Vermelha)
plt.fill_between(x, y, where=(x <= limit1), color='#e74c3c', alpha=0.45,
                 label=f'Cauda P(X <= 60g) ~ 6.68%')

# Região 2: X >= 85 g (Cauda Verde)
plt.fill_between(x, y, where=(x >= limit2), color='#2ecc71', alpha=0.45,
                 label=f'Cauda P(X >= 85g) ~ 5.21%')

# --- Marcação dos Três Percentis como Linhas Tracejadas Coloridas ---
colors_p = ['#9b59b6', '#3498db', '#e67e22'] # Roxo, Azul, Laranja
labels_p = ['P90', 'P95', 'P99']

for val, col, lab in zip(x_percentiles, colors_p, labels_p):
    plt.axvline(val, color=col, linestyle='--', lw=2.5,
                label=f'{lab} = {val:.1f} g')

# --- Marcação de Linhas Auxiliares e da Média ---
plt.axvline(limit1, color='#e74c3c', linestyle=':', lw=1.5, alpha=0.7) # Limite 60g
plt.axvline(limit2, color='#2ecc71', linestyle=':', lw=1.5, alpha=0.7) # Limite 85g
plt.axvline(mu, color='black', linestyle='-', lw=1.5, alpha=0.8, label=r'Média = 72g')

# --- Configurações do Gráfico Integrado Final ---
plt.title('Distribuição Normal Integrada (Regiões de Probabilidade e Percentis)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Massa Corporal (g)', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend(loc='upper right', frameon=True, fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Salva o gráfico
plt.savefig('grafico_normal_final_integrado.png', dpi=300)

# Mostra o gráfico (se estiver em ambiente interativo)
plt.show()
```
